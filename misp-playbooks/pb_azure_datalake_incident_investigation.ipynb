## 🔑 Step 1: Authenticate with Azure Data Lake
We use Azure credentials to access the Data Lake storage.

```python
from azure.identity import DefaultAzureCredential
from azure.storage.filedatalake import DataLakeServiceClient

# Set up authentication
account_name = 'your_account_name'
filesystem_name = 'security-logs'
credential = DefaultAzureCredential()

# Connect to Data Lake
service_client = DataLakeServiceClient(account_url=f"https://{account_name}.dfs.core.windows.net", credential=credential)
filesystem_client = service_client.get_file_system_client(filesystem_name)
print('Connected to Azure Data Lake!')
```

## 📂 Step 2: List Available Security Log Files

```python
# List log files
paths = filesystem_client.get_paths()
for path in paths:
    print(path.name)
```

## 📊 Step 3: Query Security Logs Using SQL
We will use Azure Synapse Analytics or Databricks to query logs stored in the Data Lake.

```python
import pyodbc
import pandas as pd

# Define SQL connection (replace with your connection details)
server = 'your-synapse-server.database.windows.net'
database = 'your-database'
username = 'your-username'
password = 'your-password'
driver= '{ODBC Driver 17 for SQL Server}'

conn = pyodbc.connect(f'DRIVER={driver};SERVER={server};PORT=1433;DATABASE={database};UID={username};PWD={password}')

# Define a query to detect failed login attempts
query = """
SELECT * FROM security_logs
WHERE event_type='failed_login'
AND timestamp > DATEADD(DAY, -1, GETDATE());
"""

df = pd.read_sql(query, conn)
print(df.head())
```

## 🔎 Step 4: Analyze and Interpret the Results
- Review failed login attempts.
- Correlate with other logs (e.g., privilege escalation attempts).
- Take action if suspicious behavior is detected.
